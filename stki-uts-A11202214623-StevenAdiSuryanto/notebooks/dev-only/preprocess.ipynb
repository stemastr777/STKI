```python
import re
import glob
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

### 1. Define helper functions

```python
def clean(text):
    text_cleaned = re.sub(re.compile(r"[^a-zA-Z0-9\s]"), "", text)
    text_cleaned = re.sub(re.compile(r"\d"), "", text_cleaned)
    return text_cleaned


def tokenize(text):
    words = word_tokenize(text)
    words = [word for word in words if len(words) > 1]
    words = [word.lower() for word in words]
    return words


def remove_stopwords(tokens):
    Stopwords = set(stopwords.words("indonesian"))
    return [word for word in tokens if word not in Stopwords]


def stems(tokens):
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()
    return [stemmer.stem(word) for word in tokens]
```

### 2. Load + Preprocess data

Berikut adalah tahapan preprocessing:
1. Lexical analysis (Filtering + Tokenizing + Case folding)
2. Stopword removal
3. Stemming




Tahapan **Filtering** menyebabkan data hanya berupa:
1. Huruf alphabet
2. Whitespace

```python
all_words = []
file_folder = "../data/*.txt"
for file in glob.glob(file_folder):
    print(file)
    fname = file
    file = open(file, "r", encoding="utf-8")
    text = file.read()
    cleaned_text = clean(text)

    tokens = tokenize(cleaned_text)
    tokens_no_stopwords = remove_stopwords(tokens)
    print(tokens_no_stopwords)
    tokens_stemmed = stems(tokens_no_stopwords)
    all_words.extend(tokens_stemmed)

unique_all_words = set(all_words)

unique_all_words
```
