### 1. Load + Preprocess data

```python
all_words = []
all_docs = []
all_files = []
file_folder = "../data/*.txt"
for file in glob.glob(file_folder):
    print(file)
    all_files.append(os.path.basename(file))

    file = open(file, "r", encoding="utf-8")
    text = file.read()

    cleaned_text = preprocess.clean(text)
    tokens = preprocess.tokenize(cleaned_text)
    tokens_no_stopwords = preprocess.remove_stopwords(tokens)
    tokens_stemmed = preprocess.stems(tokens_no_stopwords)
    all_words.extend(tokens_stemmed)
    all_docs.append(" ".join(tokens_stemmed))

unique_all_words = set(all_words)
```

### 2. Build Matriks TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

query = "hutan"
all_docs_query = all_docs + [query]

vectorizer = TfidfVectorizer(ngram_range=(1, 1), smooth_idf=False)
tfidf_matrix = vectorizer.fit_transform(all_docs_query)
tfidf_matrix.shape
```

```python
all_docs_query
```

### 3. Ekstrak TF-IDF query

```python
idx_query = vectorizer.vocabulary_.get(query)

print("TF-IDF " + '"' + query + '"')
tfidf_matrix[:, idx_query].toarray().ravel()
```

### 4. Hitung similarity (cosine)

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

doc_sim = cosine_similarity(tfidf_matrix)
doc_sim_df = pd.DataFrame(doc_sim)
doc_sim_df
```

### 5. Evaluasi hasil

```python
docs_similarities = doc_sim_df.iloc[all_docs_query.index(query)].values
docs_similarities
```

Ambil 3 dengan nilai cosine tertinggi **(top k-3)**

```python
import numpy as np

top_3_idxs = np.argsort(-docs_similarities)[1:4]

top_3_docs = [all_files[i] for i in top_3_idxs]
top_3_docs
```
